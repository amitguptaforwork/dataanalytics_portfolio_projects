### 2.5.1 Imputation for emp_title

- We start with a Chi-Square test.  
- A Chi-Square test is useful here because both loan_status and emp_title are categorical variables. 
- The Chi-Square test of independence determines whether there's a statistically significant relationship between two categorical variables.

````python
from scipy.stats import chi2_contingency
chi2_contingency(pd.crosstab(df.loan_status, df.emp_title))
````

<span style="font-size:15px; font-family:Arial;color:#e610ad;">
<li>p-value is 2.5 e-06.  emp_title impacts loan_status.  This feature has predictive power, so we will try to preserve its relationship with the target
<li>As they are dependent, lets check if missingness correlates with the target



- We create a binary indicator column emp_title_missing (1 = missing, 0 = not missing).
- Run chi-square between is_missing and the target.
- If is_missing is significant → treat missing as informative → impute with "Missing" category.
- If not significant → safe to replace with mode.

````python
#Check: does missingness in one col depend on another categorical col?
df['emp_title_missing'] = df['emp_title'].isna().astype(int)
ctab = pd.crosstab(df['emp_title_missing'], df['loan_status'])
chi2, p, dof, exp = chi2_contingency(ctab)
print("p-value =", p)
````

<span style="font-size:15px; font-family:Arial;color:#e610ad;">
<li>p-value is 1.6 e-135.  missing value carries some information.
<li>Thus we will impute using a new category called 'Missing' instead of Mode

- We use SimpleImputer instead of direct Pandas code (fillna) 
- SimpleImputer ensures consistent, reusable, and pipeline-friendly imputations across training, testing, and production, unlike direct pandas.fillna()

### 2.5.2 Imputation for emp_length
Approach:
- First we try to establish if this field has an impact on the target column
- Then we try to impute it.
- Again we do a Chi-Square test. emp_length, while visually numerical, is a ordinal categorical column as there are only fixed values, not real numbers.
- The Chi-Square test of independence determines whether there's a statistically significant relationship between two categorical variables.

````python
chi2, p, dof, exp = chi2_contingency(pd.crosstab(df.loan_status, df.emp_length))
print("p-value =", p)
````

<span style="font-size:15px; font-family:Arial;color:#e610ad;">
<li>p-value is 1.8 e-21.  emp_length impacts loan_status.  This feature has predictive power, so we will try to preserve its relationship with the target
<li>As they are dependent, lets check if missingness correlates with the target

````python
#Check: does missingness in one col depend on another categorical col?
df['emp_length_missing'] = df['emp_length'].isna().astype(int)
ctab = pd.crosstab(df['emp_length_missing'], df['loan_status'])
chi2, p, dof, exp = chi2_contingency(ctab)
print("p-value =", p)
````

<span style="font-size:15px; font-family:Arial;color:#e610ad;">

We are converting this categorical column to numeric one and taking care of <0 and >10 values too.

````python
import re
# Function to convert emp_length to numeric
def convert_emp_length(value):
    if pd.isna(value):
        return None
    
    value = str(value).strip()
    
    # Handle "10+ years" -> 11
    if '10+' in value:
        return 11
    
    # Handle "< 1 year" -> 0.5
    if '<' in value and '1' in value:
        return 0.5
    
    # Handle regular numbers like "4 years" -> 4
    # Extract just the number
    match = re.search(r'(\d+)', value)
    if match:
        return int(match.group(1))
    
    return None

# Apply the conversion
df['emp_length'] = df['emp_length'].apply(convert_emp_length)
````

#### drawDetailedBoxPlot
Creating a helper function to plot a boxplot that shows detailed quartile information

````python
def drawDetailedBoxPlot(data,ax):
    boxplot = ax.boxplot(data, vert=False, patch_artist=True)

    # Calculate the five-number summary
    min_val = np.min(data)
    q1 = np.percentile(data, 25)
    median = np.median(data)
    q3 = np.percentile(data, 75)
    max_val = np.max(data)

    iqr = q3-q1
    # Calculate outlier boundaries
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr


    arrowprops0 = dict(facecolor='black', shrink=0.05,mutation_scale=10)
    arrowprops1=dict(
                        arrowstyle='->',
                        color='black',
                        lw=1.8,           # Line width                     
                        mutation_scale=20 # Head size
                    ) 

    # Annotate the five-number summary on the boxplot


    ax.annotate(f'Min', xy=(min_val, 1), xytext=(min_val, 1.2),
                arrowprops=arrowprops1 , ha='center'  )

    ax.annotate(f'{min_val:.2f}', xy=(min_val, 1), xytext=(min_val, 0.87),ha='center')
                

    ax.annotate(f'Q1', xy=(q1, 1.07), xytext=(q1, 1.25),
                arrowprops=arrowprops1 , ha='center'  )
    ax.annotate(f'{q1:.2f}', xy=(q1, 1), xytext=(q1, 0.87),ha='center')


    ax.annotate(f'Median', xy=(median, 1.07), xytext=(median, 1.32),
                arrowprops=arrowprops1,ha='center'  )
    ax.annotate(f'{median:.2f}', xy=(median, 1), xytext=(median, 0.77),ha='center')

    ax.annotate(f'Q3', xy=(q3, 1.07), xytext=(q3, 1.25),
                arrowprops=arrowprops1 , ha='center')
    ax.annotate(f'{q3:.2f}', xy=(q3, 1), xytext=(q3, 0.87),ha='center')

    ax.annotate(f'Max', xy=(max_val, 1), xytext=(max_val, 1.2),
                arrowprops=arrowprops1 , ha='center')
    ax.annotate(f'{max_val:.2f}', xy=(max_val, 1), xytext=(max_val, 0.87),ha='center')

    # Add vertical dotted lines for outlier boundaries
    ax.axvline(lower_bound, color='red', linestyle='--', alpha=0.7, label=f'Lower Bound ({lower_bound:.1f})')
    ax.axvline(upper_bound, color='red', linestyle='--', alpha=0.7, label=f'Upper Bound ({upper_bound:.1f})')
    # Add vertical text labels along the lines
    ax.text(lower_bound, .7, 'Lower Bound', rotation=90, ha='center', va='bottom', color='black', fontsize=10)
    ax.text(upper_bound, .7, 'Upper Bound', rotation=90, ha='center', va='bottom', color='black', fontsize=10)
    return ax
````

#### univariate_ordinal_analysis
Creating a helper function to plot multiple graphs in one figure for ordinal data

````python
import pandas as pd
import matplotlib.pyplot as plt

def univariate_ordinal_analysis(series, column_name, order=None ):
    """
    Print summary statistics and plot distribution for an ordinal categorical Series.
    
    Parameters:
        series : pd.Series
            The ordinal categorical column.
        order : list (optional)
            Ordered list of categories (if not the Series' CategoricalDtype).
    """
    
    # Ensure categorical with order
    if not pd.api.types.is_categorical_dtype(series):
        series = series.astype("category")
    if order is not None:
        series = series.cat.set_categories(order, ordered=True)
    
    # Value counts in order
    counts = series.value_counts().reindex(series.cat.categories)
    perc = counts / counts.sum() * 100
    

    
    # Plot
    fig, ax = plt.subplots(1, 2, figsize=(10, 4))
    
    # Bar plot (counts)
    bars = counts.plot(kind="bar", ax=ax[0], color="skyblue")
    ax[0].set_title(f"{column_name} Counts by Category")
    ax[0].set_ylabel("Count")
    
    # Annotate bars with %
    for bar, pct in zip(ax[0].patches, perc):
        if not pd.isna(pct):
            ax[0].text(
                bar.get_x() + bar.get_width()/2, 
                bar.get_height(), 
                f"{pct:.1f}%", 
                ha="center", va="bottom", fontsize=9
            )
    
    # Cumulative percentage
    perc.cumsum().plot(marker="o", ax=ax[1], color="red")
    ax[1].set_title("Cumulative % by Category")
    ax[1].set_ylabel("Cumulative %")
    ax[1].set_ylim(0, 100)
    ax[1].grid(True, linestyle="--", alpha=0.5)
    
    plt.tight_layout()
    plt.show()
        # Print summary
    print(f"\n📊 Summary Statistics for {column_name}")
    print("="*50)
    print(f"Mode   : {series.mode().iloc[0]}")
    print(f"Median : {series.sort_values().iloc[len(series)//2]}")
    print("\nValue Counts (%):")
    df_analysis=pd.DataFrame({"Count": counts, "Percent": perc.round(2)})
    print(df_analysis)
    df_analysis.sort_values(by="Percent", ascending=False, inplace=True)
    print("Breakup of grades-"+", ".join([f"{idx} ({row['Percent']}%)" for idx, row in df_analysis.iterrows()]))

````

#### univariate_numerical_analysis
Creating a helper function to plot multiple graphs and other interesting information for a numerical column

````python
import scipy.stats as stats
import matplotlib.gridspec as gridspec
def univariate_numerical_analysis(data, column_name="Numerical Variable"):
    """
    Generate 4 key graphs for univariate analysis of numerical data
    """
    #Calculations
    mean = np.mean(data)
    q1= np.percentile(data,25)
    q2= np.percentile(data,50)
    q3= np.percentile(data,75)
    iqr = q3-q1
    # Calculate outlier boundaries
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    #Normality test
    if(len(data)>300):
        test_stat, p_value = stats.shapiro(data.sample(300))
    else:
        test_stat, p_value = stats.shapiro(data)

    # Create subplots
    fig = plt.figure(figsize=(10, 6))
    gs = gridspec.GridSpec(2, 3, figure=fig)
    fig.suptitle(f'Univariate Analysis: {column_name}', fontsize=16, fontweight='bold')
    
    # 1. Histogram with KDE
    ax1 = fig.add_subplot(gs[0, 0:2])  
    ax1.hist(data, bins=30, alpha=0.7, color='skyblue', edgecolor='black', density=True)
    # Add KDE curve
    x = np.linspace(data.min(), data.max(), 100)
    kde = stats.gaussian_kde(data)
    ax1.plot(x, kde(x), 'r-', linewidth=2, label='KDE')
    ax1.set_title('Distribution (Histogram + KDE)')
    ax1.set_xlabel(column_name)
    ax1.set_ylabel('Density')
    ax1.legend()
    ax1.grid(True, alpha=0.3)

    # Q-Q Plot (Normality Check)
    ax2 = fig.add_subplot(gs[0, 2])  
    stats.probplot(data, dist="norm", plot=ax2)
    ax2.set_title('Q-Q Plot (Normality Test)')
    ax2.grid(True, alpha=0.3)

    # Box Plot
    ax3 = fig.add_subplot(gs[1, 0:2])
    drawDetailedBoxPlot(data,ax3)
   
	# Clear the right subplot and add formatted text
    ax4 = fig.add_subplot(gs[1, 2]) 
    ax4.clear()
    ax4.set_xlim(0, 1)
    ax4.set_ylim(0, 1)
    ax4.axis('off')

    # Add title
    ax4.annotate('Summary Statistics', xy=(0.5, 0.95), 
                    ha='center', va='top', fontsize=12, fontweight='bold')

    #Find outliers
    lower_outliers =[]
    upper_outliers=[]
    for x in data:
        if x < lower_bound:
            lower_outliers.append(x)
        elif x > upper_bound:
            upper_outliers.append(x)
    outliers = lower_outliers + upper_outliers
    # Add statistics in a formatted way
    statistics = [
        ('Mean:', f"{mean:.2f}"),
        ('Std Dev:',f"{np.std(data):.2f}" ),
        ('Min Max:',f"({np.min(data):.2f}, {np.max(data):.2f})" ),
        ('25, 50, 75 Percentile:', f"{q1}, {q2:.2f}, {q3}" ),
        ('IQR:', iqr ),
        ('Outliers boundary:', f"({lower_bound},{upper_bound})" ),
        ('Outliers count:', f"({len(outliers)})" ),
        ('Skewness:', f"{stats.skew(data):.2f}"),
        ('Kurtosis:', f"{stats.kurtosis(data):.2f}"),
    ]
    
    if p_value<0.05:
        statistics.append(("Distribution",f"NOT Normal. Shapiro pval {p_value:.3f}"))
    else:
        statistics.append(("Distribution",f"normal Shapiro pval {p_value}"))

    print(f"Mean value is {mean:.2f} with 50% of data lying between {q1} and {q3}")      
    if(len(outliers)>0):
        print(f"There are {len(outliers)} outliers. {len(lower_outliers)} outliers are on lower side and {len(upper_outliers)} outliers are on upper side\n")  

    y_pos = 0.85
    for label, value in statistics:
        ax4.annotate(f'{label:.<20} {value}', xy=(0.1, y_pos), 
                    ha='left', va='top', fontfamily='monospace', fontsize=8)
        y_pos -= 0.08
   
    plt.tight_layout()
    plt.show()


    
````

````python
from scipy import stats

def shapiro(data):
    # --- Shapiro-Wilk test ---
    stat, p = stats.shapiro(data)
    print("Shapiro-Wilk Test: stat=%.3f, p=%.3f" % (stat, p))
    if p > 0.05:
        return("Looks normal")
    else:
        return("Not normal")
    
# Log transform (works if all data > 0)
data = df.loan_amnt
log_data = np.log(data)
print(f"log_data {shapiro(log_data)}\n")

# Square root transform
sqrt_data = np.sqrt(data)

print(f"sqrt_data {shapiro(sqrt_data)}\n")

# Box-Cox transform (requires positive data)
boxcox_data, lambda_bc = stats.boxcox(data)
print("Box-Cox lambda:", lambda_bc)

print(f"boxcox_data {shapiro(boxcox_data)}\n")

# Yeo-Johnson (works with negatives too)
from sklearn.preprocessing import PowerTransformer
pt = PowerTransformer(method='yeo-johnson')
yj_data = pt.fit_transform(data.values.reshape(-1,1)).flatten()

print(f"yj_data {shapiro(yj_data)}\n")
````

<li>Loan Amount is NOT normally distributed. We explored that even after some transformation (log, sqrt and BoxCox),it cannot be converted into Normal.
<li>Mean value is 14113.888089286165 with 50% of data lying between 8000.0 and 20000.0
<li>There are 141 outliers,ALL except 1 are 40000, it seems this is some max limit that LoanTap uses

#### Reusable helper function for ANOVA/KRUSKAL to compare categorical and numerical columns

````python
#Reusable function
#It will perform Anova or Kruksal based on input
#It will generate all groups and then perform the specific test.
# log = compact, detailed, compactNoDiagram, detailedNoDiagram

import pingouin as pg
#η² for kruskal, ε² for anova
def interpret_effect(size,metric="anova"):
    if(metric=='anova'):
        metric="ε²"
    else:
        metric="η²"

    if size < 0.01:
        effect = f"yet effect negligible ({metric}={size:.2f})."
    elif size < 0.06:
        effect = f"yet effect small ({metric}={size:.2f})."
    elif size < 0.14:
        effect = f"effect medium ({metric}={size:.2f})."
    else:
        effect = f"effect large ({metric}={size:.2f})."   
    return effect

def perform_anova_kruskal(data,numerical_column,categorical_column, test, log="detailed"):
    # Group data by category and extract numerical values
    groups = [group[numerical_column].values for _, group in data.groupby(categorical_column, observed=True)]
    p_value=0
    li="<li>"
    Ho=Ha=HaDetail=HaDetail=effectDesc=""
    

    if(test=='kruskal'):
        #Null Hypothesis= 
        Ho=f"{li}{categorical_column} and {numerical_column} are independent."
        HoDetail=f"{li} Medians of groups of {numerical_column} based on {categorical_column} are same"
        #Alternate Hypothesis
        Ha=f"{li}{categorical_column} and {numerical_column} are DEPENDENT."
        HaDetail=f"{li} Medians of groups of {numerical_column} based on {categorical_column} are different"

        # Perform Kruskal-Wallis Test
        H, p_value  = stats.kruskal(*groups)

        k = len(groups)
        n = len(data)

        # Effect sizes
        epsilon_squared = (H - k + 1) / (n - k)
        effectDesc = f" {interpret_effect(epsilon_squared,"kruskal")}"

        if(log=="detailed"):
            print(f'Ho = {Ho}')
            print(f'Ha = {Ha}')
            print(f'Kruskal-Wallis Result:{H},{p_value}')
            print(f'Effect Size:{effectDesc}')
                  

    if(test=='anova'):
        #Null Hypothesis= 
        Ho=f"{li}{categorical_column} and {numerical_column} are independent."
        HoDetail=f"{li} Mean of groups of {numerical_column} based on {categorical_column} are same"
        #Alternate Hypothesis
        Ha=f"{li}{categorical_column} and {numerical_column} are DEPENDENT."
        HaDetail=f"{li} Mean of groups of {numerical_column} based on {categorical_column} are different"

        # Perform Anova Test
        #anova_result = stats.f_oneway(*groups)
        # p_value= anova_result.pvalue

        #Using pingouin as this gives eta square value too.
        aov = pg.anova(data=data, dv=numerical_column, between=categorical_column)
        # Extract p-value and eta-squared
        p_value = aov['p-unc'].values[0]
        eta_squared = aov['np2'].values[0]  # np2 is partial eta-squared (same as eta-squared for one-way ANOVA)
        effectDesc = f" {interpret_effect(eta_squared)}"

        
        if(log=="detailed"):
            print(f'Ho = {Ho}')
            print(f'Ha = {Ha}')
            print("ANOVA Result:", p_value)     
            print(f'Effect Size (eta square):{np.round(eta_squared,2)},{effectDesc}')   

    if(p_value<0.05):
        if(log=="detailed"):
            print(f"Reject Null Hypothesis. Alternate Hypothesis '{Ha}{HaDetail}' is True, p-value:{p_value:.2f}.\n{effectDesc}")
            
        else:
            print(f'{Ha}(p:{p_value:.2f}), {effectDesc}')
                 
        
        
    else:
        if(log=="detailed"):
            print(f"Fail to Reject Null Hypothesis. Null Hypothesis '{Ho}{HoDetail}' is True, p-value:{p_value:.2f}")            
        else:
            print(f'{Ho}(p:{p_value:.2f})') 
        
    if(log=="detailed"):
        print()

    if("NoDiagram" not in log):
        # We can also do visualization
        fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(15,5))
        sns.boxplot(data=data, y=numerical_column, x=categorical_column, ax=ax[0])
        ax[0].set_ylabel(numerical_column)
        
        df_analysis = df.groupby(categorical_column, as_index=False)[numerical_column].mean()
        sns.barplot(data=df_analysis, x=categorical_column, y=numerical_column, ax=ax[1])
        ax[1].set_xlabel(categorical_column)
        ax[1].set_ylabel(numerical_column)
        ax[1].set_title(f'Average {numerical_column} by {categorical_column}')

        sns.kdeplot(data=data, x=numerical_column, hue=categorical_column, ax=ax[2])
        ax[2].set_title(f"Kdeplot for {numerical_column} by {categorical_column}")
        plt.show()
        if(log=='detailed'):
            print(df_analysis)
````

#### Reusable helper function for chi square test for comparing categorical columns

````python
from scipy.stats import chi2_contingency
def perform_chi_square_test(df, col1, col2, alpha=0.05,log="detailed"):
    """
    Perform Chi-square test on two categorical columns from a DataFrame
    
    Parameters:
    df: pandas DataFrame
    col1: string, name of first categorical column
    col2: string, name of second categorical column  
    alpha: float, significance level (default 0.05)
    
    Returns:
    dict with test results
    """
    li="<li>"
    # Create contingency table
    contingency_table = pd.crosstab(df[col1], df[col2])

    
    # Perform chi-square test
    Ho=f"{li}{col1} and {col2} are independent."
    Ha=f"{li}{col1} and {col2} are DEPENDENT"
    chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

    # Compute effect size (Cramér's V)
    n = contingency_table.sum().sum()
    phi2 = chi2 / n
    r, k = contingency_table.shape
    cramers_v = (phi2 / min(k-1, r-1)) ** 0.5

    # p-value Interpretation
    if p_value < 0.05:
        relation = "Dependent (Significant)"
    else:
        relation = "Independent (Not significant)"


    # Cramer V Effect size Interpretation
    if cramers_v < 0.1:
        effectDesc = f"association is Negligible(CramersV={cramers_v:.2f})"
    elif cramers_v < 0.3:
        effectDesc = f"association is Weak(CramersV={cramers_v:.2f})"
    elif cramers_v < 0.5:
        effectDesc = f"association is Moderate(CramersV={cramers_v:.2f})"
    else:
        effectDesc = f"association is Strong(CramersV={cramers_v:.2f})"
    
    # Results
    results = {
        'chi2_statistic': chi2_stat,
        'p_value': p_value,
        'degrees_of_freedom': dof,
        'expected_frequencies': expected,
        'contingency_table': contingency_table,
        'significant': (p_value < alpha)
    }
    
    # Print results
    if(log=="detailed"):
        print("Contingency Table:")
        print(contingency_table)
        print("\n")    
        print(f"Chi-square Test Results:")
        print(f"Chi-square statistic: {chi2_stat:.4f}")
        print(f"p-value: {p_value:.4f}")
        print(f"Degrees of freedom: {dof}")
        print(f"Significant at α={alpha}: {'Yes' if results['significant'] else 'No'}")
    
    if results['significant']:
        if(log=="detailed"):
            print(f"Reject Null Hypothesis. Alternate Hypothesis '{Ha}' is True, p-value:{results['p_value']}. However, {effectDesc}")
        else:
            print(f'{Ha}(p:{results['p_value']:.2f}), however {effectDesc}')
    else:
        if(log=="detailed"):
            print(f"Fail to Reject Null Hypothesis. Null Hypothesis '{Ho}' is True, p-value:{results['p_value']}")
        else:
              print(f'{Ho}(p:{results['p_value']:.2f})')


    # Create normalized crosstab
    crossNormalized12 = pd.crosstab(df[col1], df[col2], normalize='index') * 100
    crossNormalized21 = pd.crosstab(df[col2], df[col1], normalize='index') * 100

    # Create 1x2 subplot figure
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    colors = [
        "salmon", "skyblue",   
        "lightgreen", "gold", "orchid", "coral", "khaki",
        "plum", "turquoise", "lightsteelblue", "tan", "lightpink",
        "mediumseagreen", "wheat", "lightsalmon", "aquamarine",
        "thistle", "palegoldenrod", "powderblue", "peru", "lightcyan",
        "violet", "darkseagreen", "peachpuff", "mistyrose"
    ]
    # determine how many categories you have
    num_categories = len(crossNormalized12.columns)

    # now slice the colors list accordingly
    color_list = colors[:num_categories]
    # --- First Plot: Col2 by Col1 ---
    crossNormalized12.plot(kind='bar', stacked=False, ax=axes[0], color=color_list)
    axes[0].set_title(f"{col2} by {col1}")
    axes[0].set_ylabel(f"Percentage of {col2}")
    axes[0].set_xlabel(col1)
    axes[0].legend(title=col2)
    axes[0].tick_params(axis='x', rotation=0)

    # --- Second Plot: Col1 by Col2 ---
    crossNormalized21.plot(kind='bar', stacked=False, ax=axes[1], color=color_list)
    axes[1].set_title(f"{col1} by {col2}")
    axes[1].set_ylabel(f"Percentage of {col1}")
    axes[1].set_xlabel(col2)
    axes[1].legend(title=col1)
    axes[1].tick_params(axis='x', rotation=0)
    plt.show()

    if(log=="detailed"):
        print("="*50)    
````

### profit - engineered feature
profit earned (approx) = loan_amnt * int_rate * number of years if loan status = 1, 
profit earned (approx) = (loan_amnt * int_rate * number of years)/3 if loan status = 0.  We assume that loanee defaulted after 1/3rd of the full term (we can get this information from Loantap for better estimation) 

````python
def profitCalculation(row):
    
    if row.loan_status_code==1:        
        return  np.round(row.installment * row.term_code*12 - row.loan_amnt)
    else:
        # We assume loanee defaulted after paying 1/3rd the installments.
        return np.round(row.installment * row.term_code*12/3 - row.loan_amnt )
df['profit'] = df.apply(profitCalculation, axis=1)
````

Class that helps preserve dataframe when data is passing thru pipeline.
By default,the transformers return nd array

````python


class DataFrameWrapper(BaseEstimator, TransformerMixin):
    def __init__(self, transformer, feature_names):
        self.transformer = transformer
        self.feature_names = feature_names

    def fit(self, X, y=None):
        self.transformer.fit(X, y)
        return self

    def transform(self, X):
        X_t = self.transformer.transform(X)
        return pd.DataFrame(X_t, columns=self.feature_names, index=X.index)
````

````python
# Function to convert emp_length to numeric
def convert_emp_length(value):
    if pd.isna(value):
        return np.nan
    
    value = str(value).strip()
    
    # Handle "10+ years" -> 11
    if '10+' in value:
        return 11
    
    # Handle "< 1 year" -> 0.5
    if '<' in value and '1' in value:
        return 0.5
    
    # Handle regular numbers like "4 years" -> 4
    # Extract just the number
    match = re.search(r'(\d+)', value)
    if match:
        return int(match.group(1))   

def parse_emp_length(x):
    #x is a dataframe
    # Apply conversion function element-wise
    
    transformed = x.emp_length.apply(convert_emp_length)

    # Return as 2D array
    return transformed.to_numpy().reshape(-1, 1)   
    
    
    
features = ['emp_length']
pipeline = Pipeline(steps=[
   #('imputer', SimpleImputer(strategy='most_frequent')),
   ('parse', FunctionTransformer(parse_emp_length, validate=False)),
   ('imputer', SimpleImputer(strategy='median')),
#    ('scaler', MinMaxScaler())
])

# For a quick test, do Fit + transform and print
print(pipeline.fit_transform(X_train))

# Append to the transformers array
transformers.append((f'{features[0]}_pipe',pipeline,features))
transformed_columns.extend(features)
````

Since logistic regression is a linear model, numeric, monotonic features are ideal.

The third (months since earliest issue) works best, as it folds both year and month into one continuous value.

````python
import pandas as pd

#This is for imputation
class DateTimeFFillTransformer(BaseEstimator, TransformerMixin):
    """
    A custom transformer to convert a column to datetime and then apply forward fill.
    Assumes the input to transform is a Pandas DataFrame or Series.
    Returns a NumPy array.
    """
    def __init__(self, format='%b-%Y'):
        self.format = format # Optional: specify date format if not inferable

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        # Ensure X is a DataFrame for method chaining
        X_df = pd.DataFrame(X)
        
        # Step 1: Convert to datetime
        # errors='coerce' will turn unparseable dates into NaT (which are treated like NaN)
        for col in X_df.columns: # Handles cases where X might have multiple cols, though we target one
             X_df[col] = pd.to_datetime(X_df[col], format=self.format, errors='coerce')
        
        # Step 2: Apply forward fill
        X_filled = X_df.ffill()
        
        # Step 3: Handle potential remaining NaNs (e.g., if first value was NaN)
        X_filled = X_filled.bfill()

        
        return X_filled.values # Return as NumPy array

#This is converting issue date to a numerical value = months since earliest issued loan
def encode_issue_date(x):
    # Ensure Series
    s = x.issue_d

    # Convert to datetime
    s = pd.to_datetime(s, format='%b-%Y')
    
    # Calculate months since earliest issue
    months_since = (s.dt.year - s.dt.year.min()) * 12 + (s.dt.month - s.dt.month.min())
    
    # Return as numeric 2D array
    return months_since.to_numpy().reshape(-1, 1)

# scikit-learn compatible transformer
issue_d_pipeline = Pipeline([
    ('parse_date', FunctionTransformer(encode_issue_date, validate=False))
])


features = ['issue_d']
pipeline = Pipeline(steps=[
#    ('imputer', DateTimeFFillTransformer()),
   ('parse_date', FunctionTransformer(encode_issue_date, validate=False)),
   ('scaler', MinMaxScaler())
])

# For a quick test, do Fit + transform and print
print(pipeline.fit_transform(X_train))

#Append to the transformers array
transformers.append((f'{features[0]}_pipe',pipeline,features))
transformed_columns.extend(features)
````

### 4.5.10 earliest_cr_line and feature engineering of age_category


- Compute credit age as the difference between issue_d (loan issue date) and earliest_cr_line (first credit line date).
-This tells the customer credit agen when he was given the loan. 

- Also we will use this to guess customer age on basis of the credit age column.  
    - Minimum age: 18 years (legal requirement)
    - So if we add 18*12 to the credit age column, we may be able to guess the approximate age of customer.
    - We assume that customer has taken first loan at 18yrs of age. 

````python
# --- Custom Transformer ---
class CreditAgeAndCategoryTransformer(BaseEstimator, TransformerMixin):
    """
    Computes credit age (in months) between earliest_cr_line and issue_d,
    and creates a categorical customer_age_category from that.
    """
    def __init__(self, date_format='%b-%Y', issue_format='%b-%Y', bins=None, labels=None):
        self.date_format = date_format
        self.issue_format = issue_format
        self.bins = bins if bins is not None else [0, 30, 50, 150]
        self.labels = labels if labels is not None else ['<30', '30-50', '50+']

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        # Expect X to have two columns: [earliest_cr_line, issue_d]
        X_df = pd.DataFrame(X, columns=['earliest_cr_line', 'issue_d'])

        # Convert both to datetime
        earliest = pd.to_datetime(X_df['earliest_cr_line'], format=self.date_format, errors='coerce')
        issue = pd.to_datetime(X_df['issue_d'], format=self.issue_format, errors='coerce')

        # Compute credit age in months = (issue_d - earliest_cr_line)
        credit_age = ((issue - earliest).dt.days *12/ 365.25).fillna(0)
        credit_age[credit_age < 0] = 0  # Guard against negative durations

        # Bin into categories
        #convert credit age in months to years, then add 18 yrs to arrive at cutomer age.
        credit_age_category = pd.cut(
            credit_age/12+18,
            bins=self.bins,
            labels=self.labels,
            include_lowest=True
        ).astype(str).fillna("Unknown")

        return pd.concat([
            credit_age.rename("earliest_cr_line"),
            credit_age_category.rename("customer_age_category")
        ], axis=1)

features = ['earliest_cr_line', 'customer_age_category']
# Date pipeline: numeric age (scaled) + category (target‑encoded)
pipeline = Pipeline(steps=[
    ('calculateCreditAgeCustomerAgeCategory', CreditAgeAndCategoryTransformer(date_format='%b-%Y', issue_format='%b-%Y')),
    ('scale_encode', ColumnTransformer([
        ('CreditAge', MinMaxScaler(), [0]),        # first column = numeric age
        ('CustomerAgeCategory', TargetEncoder(), [1])  # second column = category (target‑encoded)
    ], remainder='drop'))
])


transformers.append((f'{features[0]}_pipe', pipeline, ['earliest_cr_line', 'issue_d']))
transformed_columns.extend(features)

````

### 4.5.11 Numerical columns - outlier clipping, imputation and scaling

````python
import pandas as pd
import numpy as np
from sklearn.base import BaseEstimator, TransformerMixin

class IQRBasedOutlierRemover(BaseEstimator, TransformerMixin):
    """
    IQR-based outlier clipper for numeric columns.
    Clips values outside [Q1 - factor*IQR, Q3 + factor*IQR].
    Keeps row count constant — safe for use in scikit-learn pipelines.
    """

    def __init__(self, columns=None, factor=1.5):
        """
        columns : list of columns to process (if None, all numeric columns will be used)
        factor  : multiplier for IQR (default 1.5)
        """
        self.columns = columns
        self.factor = factor
        self.bounds_ = {}

    def fit(self, X, y=None):
        """Compute IQR bounds for each column."""
        X_df = pd.DataFrame(X)
        cols_to_use = self.columns or X_df.select_dtypes(include=np.number).columns

        self.bounds_ = {}
        for col in cols_to_use:
            Q1 = X_df[col].quantile(0.25)
            Q3 = X_df[col].quantile(0.75)
            IQR = Q3 - Q1
            lower = Q1 - self.factor * IQR
            upper = Q3 + self.factor * IQR
            self.bounds_[col] = (lower, upper)
        return self

    def transform(self, X):
        """Clip each numeric column to its computed IQR bounds."""
        X_df = pd.DataFrame(X).copy()

        for col, (lower, upper) in self.bounds_.items():
            if col in X_df.columns:
                before = X_df[col].copy()
                X_df[col] = np.clip(X_df[col], lower, upper)
                changed = np.sum(before != X_df[col])
                #print(f"[IQR Clip] {col}: clipped {changed} values to range [{lower:.2f}, {upper:.2f}]")
        return X_df
    


features = ['loan_amnt','int_rate','installment','annual_inc','dti','open_acc','revol_bal','revol_util','total_acc']
pipeline = Pipeline(steps=[
    ('iqr_clipper', IQRBasedOutlierRemover(columns=features, factor=1.5)),
    ('imputer', SimpleImputer(strategy='median')),  # 2️⃣ Median imputation
    ('scaler', MinMaxScaler()),
])
transformers.append(
 ('numerical_features_pipe', pipeline, features))

transformed_columns.extend(features)
````

### 4.5.12 mort_acc
- We dont use simple imputer for mort_acc.
- We have another column total_acc and it has strong assoication with mort_acc (established during EDA)
- A model-based single imputation can be done
  - We can predict the missing values in one column using the other.
  - We are using Decision Tree Classifier (good for categorical inputs, interpretable).

````python
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier

class DecisionTreeImputer(BaseEstimator, TransformerMixin):
    """
    Uses a DecisionTree model to impute missing values in one target column
    based on one or more predictor columns.

    Parameters:
    ----------
    target_col : str
        Column with missing values to impute.
    predictor_cols : list
        Columns used to predict the missing values.
    task : {'auto', 'regression', 'classification'}, default='auto'
        Whether to use DecisionTreeRegressor or DecisionTreeClassifier.
        'auto' determines type based on target column dtype (numeric ➜ regressor, object ➜ classifier).
    random_state : int, default=42
        Random state for reproducibility.
    """

    def __init__(self, target_col, predictor_cols, task='auto', random_state=42):
        self.target_col = target_col
        self.predictor_cols = predictor_cols
        self.task = task
        self.random_state = random_state

    def fit(self, X, y=None):
        X_df = pd.DataFrame(X).copy()

        # Return self early if target col has no missing values
        if X_df[self.target_col].isna().sum() == 0:
            self.model_ = None
            return self

        # Split known/missing
        known = X_df[X_df[self.target_col].notna()]
        missing = X_df[X_df[self.target_col].isna()]

        # Choose model type automatically if not specified
        if self.task == 'auto':
            if pd.api.types.is_numeric_dtype(known[self.target_col]):
                self.model_ = DecisionTreeRegressor(random_state=self.random_state)
            else:
                self.model_ = DecisionTreeClassifier(random_state=self.random_state)
        elif self.task == 'regression':
            self.model_ = DecisionTreeRegressor(random_state=self.random_state)
        elif self.task == 'classification':
            self.model_ = DecisionTreeClassifier(random_state=self.random_state)
        else:
            raise ValueError("Invalid task. Choose 'auto', 'regression', or 'classification'.")

        # Fit the model on non-missing samples only
        if not known.empty:
            self.model_.fit(known[self.predictor_cols], known[self.target_col])

        return self

    def transform(self, X):
        X_df = pd.DataFrame(X).copy()

        if getattr(self, "model_", None) is None:
            # Nothing to impute
            return X_df.values

        missing_mask = X_df[self.target_col].isna()
        if missing_mask.any():
            preds = self.model_.predict(X_df.loc[missing_mask, self.predictor_cols])
            X_df.loc[missing_mask, self.target_col] = preds
        
        #We return only the target column, not any other columns
        return X_df[[self.target_col]].values


features = ['mort_acc']
# Build numeric preprocessing pipeline
pipeline = Pipeline([
    ('iqr_clipper', IQRBasedOutlierRemover(columns=features, factor=1.5)),
    ('decision_tree_imputer', DecisionTreeImputer(target_col='mort_acc', predictor_cols=['open_acc'])),
    ('scaler', MinMaxScaler())
])

transformers.append((f'{features[0]}_pipe', pipeline, ['mort_acc','open_acc']))

transformed_columns.extend(features)
````

### RFE (Recursive Feature Elimination)

````python
def calculate_vif(X, thresh=10.0):
    """
    Iteratively remove features with VIF above the specified threshold.
    
    Parameters:
        X (pd.DataFrame): Scaled numeric feature DataFrame
        thresh (float): VIF threshold, e.g. 10.0
        
    Returns:
        pd.DataFrame: Final VIF values for remaining features
        list: Names of features that were removed
    """
    
    variables = X.columns.tolist()
    dropped = True
    removed_features = []
    
    while dropped:
        dropped = False
        vif = pd.DataFrame()
        vif['Features'] = variables
        vif['VIF'] = [variance_inflation_factor(X[variables].values, i)
                      for i in range(len(variables))]

        max_vif = vif['VIF'].max()
        if max_vif > thresh:
            # Feature with the highest VIF
            feature_to_drop = vif.loc[vif['VIF'].idxmax(), 'Features']
            
            print(f"We can drop '{feature_to_drop}' as its VIF is ={max_vif:.2f}")
            
            variables.remove(feature_to_drop)
            removed_features.append(feature_to_drop)
            dropped = True

    # Compute final VIFs for remaining features
    final_vif = pd.DataFrame()
    final_vif['Features'] = variables
    final_vif['VIF'] = [variance_inflation_factor(X[variables].values, i)
                        for i in range(len(variables))]

    # Sort for readability
    final_vif = final_vif.sort_values(by='VIF', ascending=False).reset_index(drop=True)
    return final_vif, removed_features

final_vif, columns_to_drop = calculate_vif(X_scaled.sample(1000), thresh=10)
print("\nFinal VIF values:")
print(final_vif)

print("\nRemovable features:",len(columns_to_drop))
print(columns_to_drop)
````

Lets model the dropping of these columns as a transformer and put it in our main pipeline, just before we feed the data into the Model

````python
import pandas as pd


class ColumnDropper(BaseEstimator, TransformerMixin):
    """
    Drops specified columns from a NumPy array (output of preprocessor),
    using known column names from preprocessor.get_feature_names_out().
    """
    def __init__(self, drop_cols=None, all_feature_names=None, return_df=False):
        self.drop_cols = drop_cols or []
        self.all_feature_names = all_feature_names  # column names from preprocessor
        self.return_df = return_df  # optional: return a DataFrame if desired

    def fit(self, X, y=None):
        # Nothing to learn, just store column info
        return self

    def transform(self, X):
        # Make sure we have column names
        if self.all_feature_names is None:
            raise ValueError("all_feature_names must be provided for ColumnDropper.")
        
        # Convert numpy → DataFrame with the feature names
        df = pd.DataFrame(X, columns=self.all_feature_names)
        
        # Drop the specified columns
        df_filtered = df.drop(columns=self.drop_cols, errors='ignore')
        
        #print(f'[{self.__class__.__name__}]- Collinear columns were dropped')
        # Return DataFrame or NumPy depending on configuration
        if self.return_df:
            return df_filtered
        else:
            return df_filtered.to_numpy()

````

1. **Binary Outcome:** The dependent variable must be binary (0/1, yes/no).

4. **No Multicollinearity:** Independent variables should not be highly correlated with each other.  
5. **Large Sample Size:** The dataset should be sufficiently large to provide stable estimates.  
6. **No Perfect Separation:** No predictor should perfectly predict the outcome (i.e., all 0s or all 1s for a given X).  
7. **Independent Errors:** Residuals should not be correlated with each other.  
8. **No Strong Outliers or Influential Points:** Outliers should not unduly influence the model fit.

### 5.1.8 **No Strong Outliers or Influential Points:** Outliers should not unduly influence the model fit. 

- We use Cook’s Distance

-🔹 What Cook’s Distance Measures
-Cook’s distance (D_i) quantifies how much the overall fitted model would change if you removed observation i.

-It combines information about residual size and leverage (how far a data point’s predictors are from the mean).
-It’s a measure of a point’s influence, not just how outlying it is in terms of response.
- Intuitively: A large Dimeans: removing that point would noticeably change the model’s coefficients — i.e., it’s an influential observation.

- Rule of thumb: if Cook's D > 4/n, consider that point influential

````python
influence = model_statsmodel.get_influence()
(c, p) = influence.cooks_distance

plt.figure(figsize=(12, 5))
plt.stem(np.arange(len(c)), c, markerfmt=",")
plt.title("Cook's Distance for each observation")
plt.xlabel("Observation index")
plt.ylabel("Cook's distance")
plt.axhline(y=4/len(c), color='r', linestyle='--', label=f'Threshold: 4/n')
plt.legend()
plt.show()
````

````python
threshold = 4 / len(c)
outliers = np.where(c > threshold)[0]
print(f"Threshold: {threshold}")
print(f"Number of outliers: {len(outliers)}")
print(f"Percentage of outliers: {np.round(100*len(outliers)/len(X_scaled_statsmodel),2)}%")
print(f"Outlier indices: {outliers}")
print(f"Max Outlier to Threshold Ratio: {np.max(c)}/{threshold}={np.round(np.max(c)/threshold)}")
````

<span style="font-size:15px; font-family:Arial;color:#e610ad;">
<li>Cook’s distance is not a binary “outlier detector.” Values above the threshold don’t necessarily mean you must remove them — they simply deserve attention
<li>23,540 points above threshold,  It is 7.43% of data.= This s considered **mild**, expected for large datasets.  With ~320,000 observations, a few tens of thousands above 4/n is not alarming. ✅
<li>No super spikes in plot = Model influence distributed fairly evenly
<li>Max Cook’s D is only 6 × threshold — model seems stable ✅
